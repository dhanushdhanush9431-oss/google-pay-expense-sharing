## google pay expense sharing syatem


```python
# Google Pay Expense Sharing System

# Step 1: Create a class

class ExpenseSharing:

    # Step 2: Initialize friends

    def __init__(self, friends):

        self.friends = friends

        self.expenses = {
            friend: 0 for friend in friends
        }

    # Step 3: Add expense

    def add_expense(self, payer, amount, participants):

        if payer not in self.friends:
            print("Invalid payer")
            return

        if amount <= 0:
            print("Amount must be greater than zero")
            return

        if not participants:
            print("Participants cannot be empty")
            return

        if any(person not in self.friends for person in participants):
            print("Invalid participant")
            return

        if len(set(participants)) != len(participants):
            print("Duplicate participants are not allowed")
            return

        if payer not in participants:
            print("Payer must be a participant")
            return

        # Calculate individual share

        split_amount = amount / len(participants)

        # Update each participant balance

        for participant in participants:

            if participant != payer:

                self.expenses[participant] += split_amount

        # Update payer balance

        self.expenses[payer] -= amount - split_amount

        print("Expense added successfully!")

    # Step 4: Calculate settlement

    def calculate_settlement(self):

        print("\n----- Expense Settlement -----")

        for friend, balance in self.expenses.items():

            if balance > 0:

                print(
                    f"{friend} needs to be reimbursed "
                    f"Rs.{balance:.2f}"
                )

            elif balance < 0:

                print(
                    f"{friend} owes Rs.{-balance:.2f}"
                )

            else:

                print(f"{friend} is settled up")


# Step 5: Create friends list

friends = ["Arun", "Kumar", "Vijay"]

# Step 6: Create object

group = ExpenseSharing(friends)

# Step 7: Add expenses

group.add_expense(
    "Arun",
    900,
    ["Arun", "Kumar", "Vijay"]
)

group.add_expense(
    "Kumar",
    300,
    ["Arun", "Kumar", "Vijay"]
)

# Step 8: Calculate final settlement

group.calculate_settlement()
```
